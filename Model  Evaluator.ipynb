```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred = [1]
y_actual = [1]

if len(y_pred) != len(y_actual):
    raise ValueError("y_pred and y_actual must have the same number of values.")

print(f"Accuracy:  {accuracy_score(y_actual, y_pred):.3f}")
print(f"Precision: {precision_score(y_actual, y_pred, zero_division=0):.3f}")
print(f"Recall:    {recall_score(y_actual, y_pred, zero_division=0):.3f}")
print(f"F1 score:  {f1_score(y_actual, y_pred, zero_division=0):.3f}")
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
import matplotlib.pyplot as plt

cm = confusion_matrix(y_actual, y_pred)
ConfusionMatrixDisplay(confusion_matrix=cm).plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```
