```python
import pandas as pd
df=pd.read_csv("placements.csv")
print(df.head())
```

```python
x=df.drop(columns=["PlacedOrNot"])
y=df["PlacedOrNot"]
print(x.head())
print(y.head())
```

```python
groups=x.groupby(x.columns.tolist()).ngroup()
print(groups.head())
```

```python
print(groups.nunique())
```

```python
from sklearn.model_selection import GroupShuffleSplit
splitter=GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
```

```python
train_rows, test_row=next(splitter.split(x,y,groups=groups))
print(len(train_rows), len(test_row))
```

```python
x_train=x.iloc[train_rows]
x_test=x.iloc[test_row]
y_train=y.iloc[train_rows]
y_test=y.iloc[test_row]

```

```python
text_columns=["Gender","Stream"]
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
train_text = encoder.fit_transform(x_train[text_columns])
print(train_text[:5])
print(encoder.get_feature_names_out(text_columns))
test_text=encoder.transform(x_test[text_columns])
number_columns=["Age","Internships","CGPA", "Hostel", "HistoryOfBacklogs"]
train_numbers=x_train[number_columns]
test_numbers=x_test[number_columns]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

train_numbers_scaled = scaler.fit_transform(train_numbers)
test_numbers_scaled = scaler.transform(test_numbers)
```

```python
import numpy as np
x_train_ready = np.hstack((train_numbers_scaled, train_text))
x_test_ready = np.hstack((test_numbers_scaled, test_text))
print(x_train_ready.shape)
print(x_test_ready.shape)
```
